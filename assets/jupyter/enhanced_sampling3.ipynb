#### Sampling

As we see from the previous post, if we were able to sample from a given ensemble, we can estimate the expectation value of any observables as well as the free energy profile along any collective variable (CV). The most common ways to perform sampling are Molecular Dynamics (MD) and Monte Carlo (MC). To keep it simple, I will not talk about MD here and we will use Markov Chain Monte Carlo (MCMC) to sample from a toy 1D potential.

The idea of MCMC is to construct a Markov chain, whose equilibrium distribution $\pi(x)$ matches the target distribution $p(x)$ you want. In molecular simulations, the target distribution is the Boltzmann distribution, that is 

$$
    p(x) = \frac{\mathrm{e}^{-\beta U(x)}}{Z}   
$$

There are many ways to construct a Markov chain, a sufficient but not necessary condition is the detailed balance, which follows

$$
    p(x_i) T_{x_i\rightarrow x_j} = p(x_j) T_{x_j\rightarrow x_i}
$$

where $T_{x_i\rightarrow x_j}$ is the transition probability from $x_i$ to $x_j$.
